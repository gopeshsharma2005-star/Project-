```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.image as mpimg
```

```python
Project = pd.read_csv("C:/Users/gopes/Downloads/India_Crime_Data_States_2014_2023.csv")
Project
```

```python
crimes = ['Murder', 'Rape', 'Kidnapping', 'Assault', 'Theft', 'Robbery', 'Trafficking']
```

```python
ap_data = Project[Project['State/UT'] == 'Andhra Pradesh']
bh=Project[Project['State/UT'] == 'Bihar']
dl=Project[Project['State/UT'] == 'Delhi']
hr=Project[Project['State/UT'] == 'Haryana']
kn=Project[Project['State/UT'] == 'Karnataka']
mh=Project[Project['State/UT'] == 'Maharashtra']
```

```python
data_2014 = ap_data[ap_data['Year'] == 2014]
data_2023 = ap_data[ap_data['Year'] == 2023]
d_bh_2014 = bh[bh['Year'] == 2014]
d_bh_2023 = bh[bh['Year'] == 2023]
d_dl_2014 = dl[dl['Year'] == 2014]
d_dl_2023 = dl[dl['Year'] == 2023]
d_hr_2014 = hr[hr['Year'] == 2014]
d_hr_2023 = hr[hr['Year'] == 2023]
d_kn_2014 = kn[kn['Year'] == 2014]
d_kn_2023 = kn[kn['Year'] == 2023]
d_mh_2014 = mh[mh['Year'] == 2014]
d_mh_2023 = mh[mh['Year'] == 2023]
```

```python
ap_2014 = data_2014[crimes].values.flatten()
ap_2023 = data_2023[crimes].values.flatten()
bh_2014 = d_bh_2014[crimes].values.flatten()
bh_2023 = d_bh_2023[crimes].values.flatten()
dl_2014 = d_dl_2014[crimes].values.flatten()
dl_2023 = d_dl_2023[crimes].values.flatten()
hr_2014 = d_hr_2014[crimes].values.flatten()
hr_2023 = d_hr_2023[crimes].values.flatten()
kn_2014 = d_kn_2014[crimes].values.flatten()
kn_2023 = d_kn_2023[crimes].values.flatten()
mh_2014 = d_mh_2014[crimes].values.flatten()
mh_2023 = d_mh_2023[crimes].values.flatten()
```

```python
fig,axes=plt.subplots(3,2,figsize=(25,20))
fig.suptitle("Indian Crime Cases",fontsize=40,fontweight="bold")
img=mpimg.imread("C:/Users/gopes/Downloads/india.png")
axes[0,0].plot(crimes,ap_2014,marker="o",linestyle='-', linewidth=2, label="Year 2014", color='blue')
axes[0,0].plot(crimes, ap_2023, marker='s', linestyle='--', linewidth=2, label="Year 2023", color='red')
axes[0,0].set_title("Crime Trend in Andhra Pradesh: 2014 vs 2023", fontsize=17, fontweight='bold')
axes[0,0].set_xlabel('Crime Category')
axes[0,0].set_ylabel('Number of Reported Cases')
axes[0,0].set_facecolor("#00FFFF")
years = ['2014', '2023']
data_series = [
    [2400, 2100],   # Murder
    [3500, 4200],   # Rape
    [7000, 9500],   # Kidnapping
    [15000, 18000], # Assault
    [100000, 280000],# Theft
    [5000, 6500],   # Robbery
    [1200, 800]     # Trafficking
]
axes[0,1].stackplot(years, data_series, labels=crimes, alpha=0.8)
axes[0,1].set_title('Growth of Crime Volume Of Bihar: 2014 vs 2023', fontsize=17, fontweight='bold')
axes[0,1].set_ylabel('Cumulative Number of Cases')
axes[0,1].set_xlabel('Year')
axes[0,1].legend(loc='upper left', title="Crime Categories", bbox_to_anchor=(1, 1))
axes[0,1].set_facecolor('#3c3f41')
axes[0,1].grid(axis='y', linestyle='--', alpha=0.5)
axes[1,0].pie(hr_2014, labels=crimes, radius=1.3, 
       wedgeprops=dict(width=0.3, edgecolor='w'), pctdistance=0.95)
axes[1,0].pie(hr_2023, radius=1.0, 
       wedgeprops=dict(width=0.3, edgecolor='w'))
axes[1,0].set(aspect="equal")
axes[1,0].set_title("Comparison Of Crime Trends In Harayana",fontsize=17, fontweight="bold")
x = np.arange(len(crimes)) 
width = 0.35  
axes[1,1].bar(x - width/2, kn_2014, width, label='2014', color='skyblue', edgecolor='black')
axes[1,1].bar(x + width/2, kn_2023, width, label='2023', color='salmon', edgecolor='black')
axes[1,1].set_ylabel('Number of Reported Cases', fontweight='bold')
axes[1,1].set_title('Comparison of Crime Trends Of Karnataka: 2014 vs 2023', fontsize=17, fontweight='bold')
axes[1,1].set_xticks(x)
axes[1,1].set_xticklabels(crimes, rotation=45)
axes[1,1].yaxis.grid(True, linestyle='--', alpha=0.9)
axes[1,1].legend()
axes[1,1].set_facecolor("#B20DFF")
axes[2,0].scatter(crimes,dl_2014,label="Year 2014", color='blue')
axes[2,0].scatter(crimes,dl_2023,label="Year 2023", color='red')
axes[2,0].tick_params(axis='x', rotation=45)
axes[2,0].grid(True, linestyle='--', alpha=0.6)
axes[2,0].set_title("Crime Trend in Delhi: 2014 vs 2023", fontsize=17, fontweight='bold')
axes[2,0].set_xlabel('Crime Category')
axes[2,0].set_ylabel('Number of Reported Cases')
axes[2,0].set_facecolor("#FFB3B3")
years = ['2014', '2023']
data = np.array([mh_2014, mh_2023])
bottom = np.zeros(2) 
for i in range(len(crimes)):
    axes[2,1].bar(years, data[:, i], bottom=bottom, label=crimes[i], edgecolor='white', width=0.6)
    bottom += data[:, i] 
axes[2,1].set_ylabel('Total Number of Cases', fontweight='bold')
axes[2,1].set_title('Composition of Total Crime: 2014 vs 2023', fontsize=17, fontweight='bold')
axes[2,1].legend(title="Crime Categories", bbox_to_anchor=(1.05, 1), loc='upper left')
axes[2,1].yaxis.grid(True, linestyle='--', alpha=0.6)
```
